# Statistical analysis of abalones

```python
import pandas as pd
```

```python
header=['sex', 'length', 'diameter', 'height', 'whole weight', 
        'shucked weight', 'viscera weight', 'shell weight', 'rings']

df = pd.read_csv('data.csv')
df.columns=header
```

```python
v_count=df['sex'].value_counts()


total=sum(v_count)


v_count/=total

v_count.index=['Male','Inflant','Female']

v_count
```
